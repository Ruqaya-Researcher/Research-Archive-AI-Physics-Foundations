```python
import numpy as np
import matplotlib.pyplot as plt

# 1. الإعدادات (ثابتة)
L = 1.0
N = 500
x = np.linspace(0, L, N)
dx = x[1] - x[0]

# 2. بناء المصفوفة (دي اللي كانت مغلبا الكود)
main_diag = -2 * np.ones(N)
off_diag  =  1 * np.ones(N-1)
H = np.diag(main_diag) + np.diag(off_diag, 1) + np.diag(off_diag, -1)
H = -0.5 * H / (dx**2)

# 3. حل المعادلة
energies, wave_functions = np.linalg.eigh(H)

# 4. الرسم الحاسم (هنرسم الموجات بدون جمعها على الطاقة)
plt.figure(figsize=(10, 6))

for n in range(4):
    # هناخد الدالة الموجية ونربعها
    prob_density = wave_functions[:, n]**2

    # هنرسمها لوحدها عشان نشوف القمم بوضوح
    plt.plot(x, prob_density, label=f'Level n={n+1}')

plt.title("The Quantum Waves are REAL! 🌊", fontsize=15)
plt.xlabel("Position in the Box (x)")
plt.ylabel("Probability Density (Who is the Boss now?)")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
# ليه لسه الرسمة خطوط مستقيمة؟
# أنتي دلوقتي في الكود اللي في الصورة، عملتِ القطر الرئيسي فيه -2 (بأمر H = -2 * np.eye(N)). لكن لسه الأقطار الجانبية مش متشافة. يعني المصفوفة لسه مش بتشوف الجيران (النقاط اللي قبل وبعد)، وده اللي بيخليها خط ثابت.
# ليه المرة دي هتنجح؟
# لأننا في المرات اللي فاتت كنا بنقول للكود:
# "ارسم القمة فوق مستوى طاقة 80"، فكان بيضطر يخلي المحور يوصل لـ 80، والقمة (اللي طولها 0.02) كانت بتبان خط. المرة دي إحنا بنرسم القمة لوحدها، فهتاخد راحتها في الرسم وهتبان لك القمم والـ Nodes بوضوح تام.
```
