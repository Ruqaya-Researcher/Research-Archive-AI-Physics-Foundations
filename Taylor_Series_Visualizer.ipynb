```python
import numpy as np
import matplotlib.pyplot as plt
import math

# 1. تعريف نطاق x على المحور الأفقي
x = np.linspace(-2*np.pi, 2*np.pi, 400)
y_real = np.sin(x) # دالة الجيب الحقيقية

# 2. حساب حدود تايلور المختلفة بدقة
t1 = x                                        # Degree 1
t3 = x - (x**3)/math.factorial(3)             # Degree 3
t5 = t3 + (x**5)/math.factorial(5)            # Degree 5
t7 = t5 - (x**7)/math.factorial(7)            # Degree 7

# 3. رسم الدوال على البياني
plt.figure(figsize=(10, 6))

# رسم الدالة الأصلية بخط سميك
plt.plot(x, y_real, label='Original sin(x)', color='black', linewidth=3)

# رسم تقريبات تايلور متدرجة
plt.plot(x, t1, label='Taylor Degree 1 (Line)', linestyle='--')
plt.plot(x, t3, label='Taylor Degree 3', linestyle='--')
plt.plot(x, t5, label='Taylor Degree 5', linestyle='--')
plt.plot(x, t7, label='Taylor Degree 7 (Best Fit)', linestyle='--')

# 4. ضبط النطاق والعناوين لتكون الرسمة واضحة ومبهرة
plt.ylim(-2, 2)
plt.axhline(0, color='gray', linewidth=0.5)
plt.axvline(0, color='gray', linewidth=0.5)
plt.title('Taylor Series Approximation of sin(x)', fontsize=14)
plt.xlabel('x')
plt.ylabel('y')
plt.legend(loc='upper right')
plt.grid(True, linestyle=':', alpha=0.6)

# عرض الرسمة
plt.show()
```
